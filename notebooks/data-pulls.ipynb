```python
import matplotlib
import plotnine
import pandas as pd
import numpy as np
```

```python
raw = pd.read_csv('../data/raw/texas_tracker/20261006T023757Z/normalized.csv')
raw.head()
```

```python
texas = raw[raw['population']=='LV']
texas['margin'] = texas['dem_pct'] - texas['rep_pct']
texas.head()
```

```python
##today = pd.to_datetime('today')
set_date = pd.to_datetime('2026-10-05')
h = 14 #30#14#7 ## poll halflife
texas['age_days'] = (set_date - pd.to_datetime(texas['end_date'])).dt.days
texas['weight'] = 0.5 ** ((texas['age_days'] / h))

texas[['end_date', 'weight', 'margin', 'age_days',]].head()
```

```python
weighted_sum = 0.0
weights_sum = 0.0
for i in range(texas.shape[0]):
    weighted_sum += texas.iloc[i]['weight'] * texas.iloc[i]['margin']
    weights_sum += texas.iloc[i]['weight']
## could also use np.average(texas['margin'], weights=texas['weight']) or .sum() from pandas

weighted_avg = weighted_sum/weights_sum
print(weighted_avg)
```

```python
n_eff = (texas['weight']).sum()**2 / (texas['weight']**2).sum()
print(n_eff)
```

```python
def weighted_avg(polls, ref_date, halflife):
    polls = polls[pd.to_datetime(polls['end_date'])<= ref_date]
    if polls.empty:
        print("No polls found in this time period")
        return None, None
    polls['age_days'] = (ref_date - pd.to_datetime(polls['end_date'])).dt.days
    polls['weight'] = 0.5 ** ((polls['age_days'] / halflife))
    avg = np.average(polls['margin'], weights=polls['weight'])
    n_eff = (polls['weight']).sum()**2 / (polls['weight']**2).sum()
    print(f"Average margin: {avg:.2f} (n_eff: {n_eff:.2f})")
    return avg, n_eff
```

```python
weighted_avg(texas, set_date, h)
```

```python
from scipy.stats import norm

def p_dem_win(avg, sigma):
    prob = norm.cdf(avg / sigma)
    print(f"Probability of a Democrat winning: {prob * 100:.2f}%")
    return prob
```

Create add margins functions for the two types of results

```python
def add_margins_polls(raw_poll):
    raw_poll['margin'] = raw_poll['dem_pct'] - raw_poll['rep_pct']
    return raw_poll

def add_margins_results(wide_results):
    wide_results['margin'] = wide_results['dem_pct'] - wide_results['rep_pct']
    return wide_results
```

Incorporate and clean FEC/MIT Historical Election Results

```python
base = '../data/processed/senate_calibration/20261006T024116Z_20261006T035630Z_v1'

results = pd.read_csv(f'{base}/candidate_results.csv',
                      dtype=str, keep_default_na=False)

results.head()
```

```python
rows = results[
    (results['contest_status'] == 'eligible') &
    (results['valid_vote'] == 'true')
].copy()

rows['votes'] = rows['votes'].astype(int)

rows.head()
```

```python
wide = rows.pivot_table(
    index='contest_id',
    columns='side',
    values='votes',
    aggfunc='sum',
    fill_value=0
).rename(columns={'D':'dem_votes', 'R':'rep_votes', 'other':'other_votes'})

wide.head()

wide['valid_votes'] = wide.sum(axis=1)
wide['dem_pct'] = wide['dem_votes'] / wide['valid_votes'] * 100
wide['rep_pct'] = wide['rep_votes'] / wide['valid_votes'] * 100

wide = (
    wide.rename_axis(columns='election')
)

wide = add_margins_results(wide)

wide.to_csv('../outputs/results_wide.csv', index=True, mode='w', sep=',')
wide.head()
```

Incorporate and clean historical polls

```python
base = '../data/processed/senate_calibration/20261006T024116Z_20261006T035630Z_v1'

poll_results = pd.read_csv(f'{base}/poll_candidate_rows.csv', dtype=str, keep_default_na=False)

poll_results.head()
```

```python
poll_rows = poll_results[poll_results['question_status'] == 'selected'].copy()

poll_rows['pct'] = poll_rows['pct'].astype(float)

poll_rows.head()
```

```python
wide_polls = poll_rows.pivot_table(
    index='question_key',
    columns='side',
    values='pct',
    aggfunc='sum',
    fill_value=0
).rename(columns={'D':'dem_pct', 'R':'rep_pct', 'other':'other_pct'})

wide_polls = add_margins_polls(wide_polls).reset_index()

poll_questions = pd.read_csv(f'{base}/poll_questions.csv', dtype=str, keep_default_na=False)

wide_polls = wide_polls.merge(poll_questions, on='question_key',
                              how='left', validate='one_to_one')

wide_polls.to_csv('../outputs/polls_wide.csv', index=False, mode='w', sep=',')
wide_polls.head()
```
